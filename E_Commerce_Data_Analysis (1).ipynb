# E-Commerce Sales and Customer Analytics

**Problem Statement**

An e-commerce business wants to understand its customer purchasing behavior, product performance, and overall sales patterns. The dataset contains customer demographics, product details, pricing and discounts, and order-level transaction information. The objective of this Exploratory Data Analysis (EDA) is to identify important patterns and relationships in customer behavior, product sales, revenue, discounts, ratings, order trends, and geographical performance. These insights can help the business understand its current performance and identify areas that may require further attention.**



**Objectives**

** 1. Customer Analysis**

* Understand customer demographics and age groups
* Analyze customer tiers
* Identify customer distribution by city/state
* Analyze total orders and total spending
* Understand high-value customer segments

**2. Product Analysis**

* Analyze product categories and brands
* Examine original price vs selling price
* Study discount patterns
* Analyze product ratings and reviews
* Identify products with high/low sales performance

**3. Sales Analysis**

* Analyze order volume and revenue
* Examine monthly/weekly sales trends
* Analyze order value and quantity
* Study payment modes and order statuses
* Analyze geographical sales patterns

**4. Customer Experience**

* Analyze ratings and review patterns
* Examine ratings across categories/products
* Investigate the relationship between discounts and ratings
* Analyze review availability

**5. Relationships & Patterns**

* Customer age vs spending
* Discount vs quantity sold
* Product price vs quantity sold
* Rating vs sales
* Shipping cost vs order value
* Customer tier vs spending

# Detecting outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_cols = ['Age', 'Total_Orders', 'Total_Spent']

for col in numerical_cols:
    plt.figure(figsize=(8, 5))

    sns.histplot(c_data[col].dropna(), bins=30, kde=True)

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

Since the distribution of these columns are skewed we use IQR method to detect outliers

```python
num_cols = ['Age', 'Total_Orders', 'Total_Spent']

for col in num_cols:
    Q1 = c_data[col].quantile(0.25)
    Q3 = c_data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = c_data[(c_data[col] < lower_limit) | (c_data[col] > upper_limit)]

    print(f"\nColumn: {col}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"Lower Limit: {lower_limit}")
    print(f"Upper Limit: {upper_limit}")
    print(f"Number of Outliers: {len(outliers)}")
```

```python
# Define the IQR limits
limits = {
    'Age': (0.5, 68.5),
    'Total_Orders': (-1.5, 10.5),
    'Total_Spent': (-160402.2575, 368174.8225)
}

# Display outlier values for each column
for col, (lower, upper) in limits.items():

    outliers = c_data[(c_data[col] < lower) | (c_data[col] > upper)]

    print(f"\nColumn: {col}")
    print("Outlier values:")
    print(outliers[col].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_cols = [
    'Quantity',
    'Unit_Price',
    'Order_Value',
    'Shipping_Cost',
    'Coupon_Discount'
]

for col in numerical_cols:
    plt.figure(figsize=(8, 5))

    sns.histplot(s_data[col].dropna(), bins=30, kde=True)

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

All the distributions are **Skewed**

```python
#  IQR method is used since the distribution of data is skewed

numerical_cols = [
    'Quantity',
    'Unit_Price',
    'Order_Value',
    'Shipping_Cost',
    'Coupon_Discount'
]

for col in numerical_cols:
    Q1 = s_data[col].quantile(0.25)
    Q3 = s_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = s_data[
        (s_data[col] < lower_limit) |
        (s_data[col] > upper_limit)
    ]

    print(f"\n{col}")
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
```

```python
# Define the IQR limits
limits = {
    'Quantity': (1.0, 1.0),
    'Unit_Price': (-32319.48, 57995.08),
    'Order_Value': (-37958.655, 68124.185),
    'Shipping_Cost': (0.0, 0.0),
    'Coupon_Discount': (0.0, 0.0)
}

# Display outlier values for each column
for col, (lower, upper) in limits.items():

    outliers = s_data[
        (s_data[col] < lower) |
        (s_data[col] > upper)
    ]

    print(f"\nColumn: {col}")
    print("Outlier values:")
    print(outliers[col].value_counts())
```

# Exploratory Data Analysis

```python
# Numerical Variable Analysis

numerical_cols = [
    'Quantity',
    'Unit_Price',
    'Order_Value',
    'Shipping_Cost',
    'Coupon_Discount',
    'Total_Amount',
    'Rating',
    'Customer_Age'
]

for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):

        x_col = numerical_cols[i]
        y_col = numerical_cols[j]

        plt.figure(figsize=(8, 5))

        sns.scatterplot(
            data=s_data,
            x=x_col,
            y=y_col
        )

        plt.title(f'{x_col} vs {y_col}')
        plt.xlabel(x_col)
        plt.ylabel(y_col)

        plt.tight_layout()
        plt.show()
```

**Conclusion**

The scatter plot shows that Coupon_Discount does not have a clear relationship with Rating.

* Ratings are concentrated at 3.0, 4.0, and 5.0.
* All three rating levels occur across a wide range of coupon discounts.
* 5-star ratings appear most frequently and are spread across both low and high discounts.
* 4-star ratings are also distributed across the discount range, but with fewer observations.
* 3-star ratings are mainly concentrated at lower discount values, with only a few observations at very high discounts.
* Overall, increasing the coupon discount does not consistently lead to higher ratings.

Dashboard insight: Coupon discount appears to have limited influence on customer ratings. It would be useful to analyze it alongside variables such as destination, passenger type, age, travel distance, or coupon type to identify stronger factors affecting ratings.


```python
# Examinig the corelation between numerical variables

numerical_cols = [
    'Quantity',
    'Unit_Price',
    'Order_Value',
    'Shipping_Cost',
    'Coupon_Discount',
    'Total_Amount',
    'Rating',
    'Customer_Age'
]

plt.figure(figsize=(10, 7))

corr = s_data[numerical_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

**Conclusion**

The correlation analysis shows that Unit_Price has a strong positive relationship with Order_Value and Total_Amount (0.88), while Order_Value and Total_Amount are perfectly correlated (1.00), indicating potential redundancy. Quantity and Coupon_Discount have weak positive relationships with transaction values. Rating and Customer_Age show little linear relationship with the financial variables. Shipping_Cost also has relatively weak correlations with most variables.
